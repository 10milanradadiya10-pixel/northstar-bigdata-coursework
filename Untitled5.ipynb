```python
import sqlite3

conn = sqlite3.connect("northstar.db")

customers.to_sql(
    "customers",
    conn,
    if_exists="replace",
    index=False
)

print("SQL table created")
```

```python
import pandas as pd

query = """
SELECT customer_type, COUNT(*) as total_customers
FROM customers
GROUP BY customer_type
"""

result = pd.read_sql(query, conn)

result
```

```python
orders = pd.read_csv('northstar_dataset/northstar_dataset/orders.csv')

orders.to_sql("orders", conn, if_exists="replace", index=False)

print("Orders loaded into SQL")
```

```python
import pandas as pd

query = """
SELECT c.customer_type, COUNT(o.order_id) as total_orders
FROM customers c
JOIN orders o
ON c.customer_id = o.customer_id
GROUP BY c.customer_type
"""

result = pd.read_sql(query, conn)

result
```

```python
import matplotlib.pyplot as plt

result.plot(kind="bar", x="customer_type", y="total_orders")

plt.title("Total Orders by Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Total Orders")

plt.show()
```

```python
deliveries = pd.read_csv('northstar_dataset/northstar_dataset/deliveries.csv')

deliveries.to_sql("deliveries", conn, if_exists="replace", index=False)

print("Deliveries loaded")
```

```python
deliveries["dispatch_time"] = pd.to_datetime(deliveries["dispatch_time"])
deliveries["delivery_completed_at"] = pd.to_datetime(deliveries["delivery_completed_at"])

deliveries["delivery_duration"] = (
    deliveries["delivery_completed_at"] - deliveries["dispatch_time"]
).dt.total_seconds() / 60

deliveries["delivery_duration"].describe()
```

```python
cleaned_deliveries = deliveries[deliveries["delivery_duration"] > 0]
```

```python
import matplotlib.pyplot as plt

plt.hist(cleaned_deliveries["delivery_duration"], bins=20)

plt.title("Delivery Time Distribution")
plt.xlabel("Minutes")
plt.ylabel("Number of Deliveries")

plt.show()
```

```python
import mongomock

client = mongomock.MongoClient()

db = client["northstar_db"]

print("MongoDB database created")
```

```python
import pandas as pd

# Load datasets
customers = pd.read_csv(
    'northstar_dataset/northstar_dataset/customers.csv'
)

complaints = pd.read_csv(
    'northstar_dataset/northstar_dataset/complaints.csv'
)

# Convert to MongoDB document format
customer_records = customers.to_dict("records")
complaint_records = complaints.to_dict("records")

# Create collections
customers_collection = db["customers"]
complaints_collection = db["complaints"]

# Insert documents
customers_collection.insert_many(customer_records)
complaints_collection.insert_many(complaint_records)

print("MongoDB collections created successfully")
```

```python
print("Total customers:", customers_collection.count_documents({}))
print("Total complaints:", complaints_collection.count_documents({}))
```

```python
new_customer = {
    "customer_id": "C9999",
    "age": 30,
    "home_zone": "Central",
    "customer_type": "Consumer",
    "loyalty_score": 88.5,
    "account_status": "Active"
}

customers_collection.insert_one(new_customer)

print("New customer inserted")
```

```python
customers_collection.find_one(
    {"customer_id": "C9999"},
    {"_id": 0}
)
```

```python
customers_collection.update_one(
    {"customer_id": "C9999"},
    {"$set": {"loyalty_score": 95.0}}
)

customers_collection.find_one(
    {"customer_id": "C9999"},
    {"_id": 0}
)
```

```python
customers_collection.delete_one(
    {"customer_id": "C9999"}
)

customers_collection.find_one(
    {"customer_id": "C9999"},
    {"_id": 0}
)
```

```python
customers_collection.delete_one(
    {"customer_id": "C9999"}
)

result = customers_collection.find_one(
    {"customer_id": "C9999"},
    {"_id": 0}
)

print(result)
```
